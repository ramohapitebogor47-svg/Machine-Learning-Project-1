## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

A fixed rule (like flag evry page older than 180 days) is too simple.It treats a page with 5 views the exact same as page with 50 000 views.
The pattern is too messy for a simple if-statement because a page's health depends on many things happening at once.A page might be old,but if it has a high CTR and is in 1st position, its still healthy.ML can look at all these numbers together to find complex combinations that a human would miss.
The model outputs ranked a list of the top 50 pages.The content team uses this list to decide exactly which pages to rewrite first, making sure they don't waste time on pages that are already doing fine.
