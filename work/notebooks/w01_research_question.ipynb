## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

DATA_URL = "https://raw.githubusercontent.com/ramohapitebogor47-svg/Machine-Learning-Project-1/main/data/raw/content_refresh_anonymized.csv"

#1.load starter data
df = pd.read_csv(DATA_URL)

#2.Calculate the real numbers
total_pages = len(df)
declining_pages = len(df[df["trend_direction"]=="down"])
median_stale = df["days_since_last_update"] .median()
stale_visible = len(df[(df["days_since_last_update"]> 180) & (df["impressions_90d"] > 100)])

#3.Print the results
print(f"REAL NUMBER 1: Out of {total_pages} total pages, {declining_pages} are currently declining.")
print(f"REAL NUMBER 2: The median page hasn't been updated in {median_stale} days.")
print(f"REAL NUMBER 3: There are {stale_visible} pages that are stale (>180 days) but still visible(>100 impressions).")
```

The data confirms a large opportunity.The median page hasn't been updated in 20 days,and there are 35 pages that are both stale and visible.This proves there is a measurable backlog of content that needs prioritizing.
